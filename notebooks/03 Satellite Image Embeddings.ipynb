```python
def fetch_image(lat, lon, id_value, size=256):

    bbox = BBox(bbox=[lon-0.001, lat-0.001, lon+0.001, lat+0.001], crs=CRS.WGS84)
    
    request = SentinelHubRequest(
        data_folder="../images",
        evalscript="""
        // RGB output
        return [B02, B03, B04];
        """,
        input_data=[SentinelHubRequest.input_data(data_collection=DataCollection.SENTINEL2_L2A)],
        responses=[SentinelHubRequest.output_response("default", MimeType.PNG)],
        bbox=bbox,
        size=(size, size),
        config=config
    )

    result = request.get_data(save_data=True)
```

```python
for index, row in df.head(10).iterrows():        # starting with 10 only
    lat = row['lat']
    lon = row['long']
    pid = row['id']

    print("Downloading:", pid)

    try:
        fetch_image(lat, lon, pid)
    except Exception as e:
        print("FAILED:", pid, e)
```
