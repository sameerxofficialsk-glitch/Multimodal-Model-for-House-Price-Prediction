```python
##TRAINING & EVALUATING BASELINE MODELS 
```

```python
train_df = pd.read_csv("train(1)(train(1)).csv")
test_df = pd.read_csv("test2(test(1)).csv")

for df in [train_df, test_df]:
    df['zipcode'] = df['zipcode'].astype(str)
    df.drop(columns=['date'], inplace=True)

X_train = train_df.drop(columns=['price', 'id', 'zipcode'])
y_train = train_df['price']

X_test = test_df.drop(columns=['id', 'zipcode'])
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=300,
    max_depth=25,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
rf_test_preds = rf.predict(X_test)
```

```python
submission = pd.DataFrame({
    "id": test_df["id"],
    "predicted_price": rf_test_preds
})

submission.to_csv("rf_predictions_final.csv", index=False)
```

```python
## Because the test dataset does not contain actual price values.
## So we cannot compute RMSE/MAE.
## Instead, we evaluate your model on the train dataset using cross-validation
```

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    rf, X_train, y_train, cv=5, scoring='neg_mean_absolute_error'
)

print("MAE CV Score:", -cv_scores.mean())
```

```python
cv_r2 = cross_val_score(
    rf, X_train, y_train, cv=5, scoring='r2'
)
print("R2 CV Score:", cv_r2.mean())
```
